## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Unit of Analysis: A unique page (page_id) per client over a 90-day trajectory.

The Core Question: Which high-value pages are showing early, statistically reliable indicators of performance decline?

The Decision: Deciding whether to allocate content refreshing resources (writers, editors, technical SEO) to a specific URL.

The Action: Queueing prioritized pages into a content refresh workflow to update outdated statistics, refresh headings, and optimize internal linking.

Cost of a Wrong Call:

False Positive (Misidentifying a stable page as declining): Wastes client content budget and human editing effort on a page that didn't need intervention.

False Negative (Missing a genuinely declining page): Allows key high-converting search pages to lose ranking positions and organic traffic permanently to competitors.

Why Machine Learning Helps: ML models can evaluate multi-dimensional decay patterns (combining position changes, CTR variance, impression drops, and content age) faster and more consistently than simple static rules.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
import numpy as np
import os

# Path to the dataset
file_path = '../../data/raw/starter_dataset.csv'

if os.path.exists(file_path):
    df = pd.read_csv(file_path)
    data_source = "Real"
else:
    # Fallback to synthetic data if file is missing to prevent execution errors
    data = {
        'page_id': range(1, 101),
        'is_declining_label': np.random.choice([0, 1], size=100, p=[0.7, 0.3]),
        'content_age_days': np.random.randint(30, 700, size=100)
    }
    df = pd.DataFrame(data)
    data_source = "Synthetic"

# Number 1: Total volume of pages evaluated
total_pages = len(df)

# Number 2: Percentage of pages flagged with declining traffic trends
declining_pages = df['is_declining_label'].sum()
declining_pct = (declining_pages / total_pages) * 100

# Number 3: Average content age (in days) for declining vs. non-declining pages
avg_age_declining = df[df['is_declining_label'] == 1]['content_age_days'].mean()
avg_age_stable = df[df['is_declining_label'] == 0]['content_age_days'].mean()

print(f"--- Starter Data Discovery Numbers ({data_source}) ---")
print(f"1. Total Dataset Size: {total_pages:,} pages evaluated across clients.")
print(f"2. Declining Page Ratio: {declining_pages:,} pages ({declining_pct:.2f}%) exhibit performance decay.")
print(f"3. Content Age Gap: Declining pages average {avg_age_declining:.1f} days old vs. {avg_age_stable:.1f} days for stable pages.")
```
